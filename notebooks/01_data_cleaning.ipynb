```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
PROJECT_ROOT = Path.cwd().parent
RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

print(PROJECT_ROOT)
print(RAW_DIR)
```

Path.cwd()：获取当前notebook运行时的工作目录。Jupyter notebook的"当前目录"默认是notebook文件所在位置，也就是nhaven-higher-ed-analysis/notebooks/。

parent：往上退一级，回到项目根目录nhaven-higher-ed-analysis/。这样无论你以后从哪里启动Jupyter，只要notebook还在notebooks/文件夹里，这行代码都能正确定位到根目录——不依赖你电脑上的绝对路径（比如/Users/gigi/Desktop/...），这对以后要不要分享给别人看/上传GitHub都很重要。

RAW_DIR/PROCESSED_DIR：提前把两个文件夹路径定义成变量，而不是每次读写文件都手打路径字符串。好处是如果以后文件夹结构变了，只改这两行，不用满notebook找字符串改。

最后两个print是验证性代码，不是最终逻辑的一部分——运行完看看打印出来的路径对不对，确认没跑偏，之后可以留着也可以删掉。

```python
raw_files = list(RAW_DIR.glob("*.csv"))
print(raw_files)
```

RAW_DIR.glob("*.csv")：glob是"模式匹配"，*.csv表示"匹配所有以.csv结尾的文件"。这样写而不是直接手打文件名，是因为文件名里带了日期（Data_7-3-2026.csv），以后你重新导出数据、文件名一变，硬编码的文件名就得手动改。用glob扫描文件夹，代码更"活"，不依赖具体文件名。

list(...)：glob返回的是一个生成器（generator），不是直接能看到内容的东西，包一层list()才能打印出实际的文件路径列表。

```python
raw_path = raw_files[0]
df = pd.read_csv(raw_path)

print(df.shape)
df.head()
```

raw_files[0]：上一步raw_files是个列表（哪怕只有一个文件，glob返回的也是列表）。因为你现在data/raw/里只有这一个CSV，所以直接取第0个元素就是它。注意：以后如果你往raw/文件夹里放了不止一个CSV（比如又下载了新版本忘记删旧的），这行代码会出问题——它永远只读列表里第一个，可能不是你想要的那个。现阶段先这样写没问题，但养成习惯：以后raw/文件夹里只放"当前正在用"的那一份数据，或者以后我们可以改成更精确地指定文件名。

pd.read_csv(raw_path)：pandas读取CSV的标准函数，返回一个DataFrame对象（你可以理解成Python里的Excel表格）。Path对象可以直接传给read_csv，不需要转成字符串，pandas自己认得。

df.shape：打印(行数, 列数)。你之前我帮你看过这份数据，应该是(4, 39)——4所学校，39列（包括那个全空的"Unnamed: 38"）。这一步是用来验证读取有没有出问题——如果行数列数跟预期差很远（比如变成1行或者上百列），说明CSV的分隔符、编码、表头行数不对，需要调整read_csv的参数。

df.head()：显示前5行数据（这里数据总共才4行，所以会全部显示）。这是目视检查——看看列名有没有乱码、数字有没有被读成奇怪的格式、有没有整行整列错位。notebook里df.head()作为cell最后一行，会自动渲染成好看的表格，不需要额外套print()。

```python
df = df.drop(columns=["Unnamed: 38"])
df.shape
```

```python
for col in df.columns:
    print(col)
```

```python
rename_map = {
    "UnitID": "unit_id",
    "Institution Name": "institution",
}

years = [2023, 2021, 2019, 2017, 2015, 2013]

for year in years:
    rename_map[f"4-year Graduation rate - bachelor's degree within 100% of normal time (GR200_{str(year)[-2:]}_RV)"] = f"grad_rate_4yr_{year}"
    rename_map[f"6-year Graduation rate - bachelor's degree within 150% of normal time (GR200_{str(year)[-2:]}_RV)"] = f"grad_rate_6yr_{year}"
    rename_map[f"Part time total (EF{year}_RV  All students  Undergraduate total)"] = f"enroll_pt_{year}"
    rename_map[f"Full time total (EF{year}_RV  All students  Undergraduate total)"] = f"enroll_ft_{year}"
    rename_map[f"Full-time retention rate  {year} (EF{year}D_RV)"] = f"retention_ft_{year}"
    rename_map[f"Part-time retention rate  {year} (EF{year}D_RV)"] = f"retention_pt_{year}"

print(len(rename_map))
```

f"..."是Python的f-string语法，{year}会被替换成变量的实际值。比如year=2023时，f"grad_rate_4yr_{year}"就变成字符串"grad_rate_4yr_2023"。

str(year)[-2:]：把年份2023转成字符串"2023"，再取最后两位"23"——这是因为IPEDS原始列名里的编码是两位数（GR200_23_RV），不是完整四位年份。

```python
missing = [col for col in rename_map.keys() if col not in df.columns]
print(missing)
```

```python
df = df.rename(columns=rename_map)
df.columns.tolist()
```

```python
long_df = pd.wide_to_long(
    df,
    stubnames=["grad_rate_4yr", "grad_rate_6yr", "enroll_pt", "enroll_ft", "retention_ft", "retention_pt"],
    i=["unit_id", "institution"],
    j="year",
    sep="_"
)

long_df = long_df.reset_index()
print(long_df.shape)
long_df.head(10)
```

stubnames：告诉函数"这6个是指标名的词根（stub），你要在列名里找这些前缀，后面跟着的下划线+数字就是要拆出来的年份"。顺序不影响结果，但我保持了和你数出来一致的6个。

i=["unit_id", "institution"]：这是"每一行的身份证"——转换后，这两列的组合用来标识"这是哪所学校"，不会被拆开，保留原样。

j="year"：新生成的、用来存放"年份"的列，取名叫year。
sep="_"：前面解释过，告诉函数指标名和年份之间的分隔符是下划线。

pd.wide_to_long()执行完，返回的结果默认把i和j指定的列变成索引（index），而不是普通列，这个是 wide_to_long比较反直觉的地方，新手很容易漏掉这一步，导致后面unit_id/year用不了。所以必须紧接着调用.reset_index()，把它们从索引"还原"成普通列，才能正常用df["year"]这样取值。

print(long_df.shape)：验证转换结果。我们预期应该是(24, 9)——4所学校×6年=24行，9列（unit_id, institution, year + 6个指标）。这是刚才你数出"6种指标"派上用场的地方,用来验证转换对不对。
